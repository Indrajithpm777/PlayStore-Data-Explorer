### New Columns

```python
apps['released_year']=apps['released'].dt.year.astype('Int64')
# Extracted the year from the released column using .dt.year and created a new column released_year, casting its data type to Int64 to properly support and handle null values.
```

```python
apps['last_updated_year']=apps['last_updated'].dt.year
# extracted last updated year from the date. no nulls in the released column so dtype will be int.
```

```python
apps['app_age']=apps['last_updated_year']-apps['released_year'].astype('Int64')
# created this column for knowing the age of the application, its the diffrence between released_year and released.also changed dtype to Int64(nullable intiger data type)because its having nulls.
```

```python
def categorize_monetization(row):
    if not row['free']:
        return 'Paid'
    else:
        has_ads = row['contains_ads']
        has_iap = row['offers_iap']
        if has_ads and has_iap:
            return 'Free with Ads and IAP'
        elif has_ads:
            return 'Free with Ads'
        elif has_iap:
            return 'Free with IAP'
        else:
            return 'Free without Ads and IAP'
    
```

```python
apps['monetization_category']=apps.apply(categorize_monetization,axis=1)
# monetization_category was created in the dataframe to classify each app based on whether it is paid, completely free, or free with advertisements and in-app purchases using created function.
```
